### In Synthetic data

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
````

````python
X, y = make_classification(n_samples = 500 , n_features = 2 , n_redundant = 0, n_informative = 2, n_clusters_per_class = 1, n_classes = 2, random_state = 8)

_, ax = plt.subplots(figsize=( 5 , 4 ))
ax.scatter( X[:, 0] , X[:, 1] , c = y , cmap = 'Paired', edgecolors="k")
ax.set_axis_off()
````

````python
# Create an instance
logistic_regression = LogisticRegression()

# Train the model
logistic_regression.fit ( X , y ) #(# Samples, # Features)
````

````python
logistic_regression.coef_
````

````python
X[0,]
````

````python
yn = (logistic_regression.coef_[0, 0] * X[0] + logistic_regression.coef_[0, 1] * X[0]) + logistic_regression.intercept_
````

````python
print(yn.shape)
````

### Decision boundaries

````python
# Min-Max values for the grid of decision boundaries
x_min, x_max = np.min(X[:, 0]) -.5 , np.max(X[:, 0]) +.5
y_min, y_max = np.min(X[:, 1]) -.5 , np.max(X[:, 1]) +.5

# Create grid
x_grid, y_grid = np.meshgrid( np.linspace(x_min, x_max, 200) , np.linspace(y_min, y_max, 200) )

# Estimate predictions over the grid
predictions = logistic_regression.predict(np.c_[x_grid.ravel(), y_grid.ravel()])
print (predictions.shape)
````

````python
# Reshape predictions
predictions = predictions.reshape(x_grid.shape)

# Draw grid
_, ax = plt.subplots (figsize=(8,5))
ax.pcolormesh( x_grid , y_grid , predictions , cmap = "Paired")

# Scatter real values
ax.scatter ( X[:, 0] , X[:, 1] , c = y , edgecolors = 'k', cmap = "Paired")
ax.scatter ( X[0], yn )

ax.set_xticks(())
ax.set_yticks(())
````
