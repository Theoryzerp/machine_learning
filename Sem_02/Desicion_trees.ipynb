## Creating Synthetic data

````python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import make_classification
````

````python
X, y = make_classification(n_samples = 500 , n_features = 2 , n_redundant = 0, n_informative = 2, n_clusters_per_class = 1, n_classes = 2, random_state = 8)

_, ax = plt.subplots(figsize=( 5 , 4 ))
ax.scatter( X[:, 0] , X[:, 1] , c = y , edgecolors = "k", cmap = 'Paired')
ax.set_xticks([])
ax.set_yticks([])
````

## Splitting nodes in a decision tree?

A classification tree tries to create groups that are as pure as possible.

### Gini impurity

$$Gini = 1 - \sum_{i=1}^{k} p_i^2$$

### Entropy

$$Entropy = -\sum_{i=1}^{k} p_i \log_2(p_i)$$

Here, $p_i$ is the proportion of observations belonging to class $i$. Lower impurity means a more homogeneous node.


````python
# Create an instance
decision_tree = DecisionTreeClassifier(criterion="gini", max_depth=5)

# Train the model
decision_tree.fit ( X, y ) #(# Samples, # Features)
````

## Visualizing the Decision Tree

````python
_, ax = plt.subplots(figsize=(12,10))
plot_tree(decision_tree, filled=True, fontsize=5, ax=ax)
````

### How to read a node

- **Condition**: the question used to split the data.
- **gini**: node impurity.
- **samples**: training observations in the node.
- **value**: observations from each class.
- **class**: predicted class.

## Load the Iris Classification Dataset

The [Iris dataset](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_iris.html#sklearn.datasets.load_iris) contains measurements for three flower species: Setosa, Versicolor, and Virginica.

````python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.datasets import load_iris
````

````python
iris = load_iris() # Load iris dataset
iris_X = iris.data # Features
iris_y = iris.target # Labels

# Imprimir formas
print (iris_X.shape, iris_y.shape)
````

````python
# Create an instance
decision_tree = DecisionTreeClassifier(criterion= "gini")

# Train the model
decision_tree.fit ( iris_X , iris_y ) #(# Samples, # Features)
````

````python
_, ax = plt.subplots(figsize=( 12 , 10 ))
plot_tree(decision_tree, filled=True, fontsize=5, ax=ax)
````

## Advantages and Limitations

### Advantages

- Easy to understand and visualize.
- Requires little data preparation.
- Handles nonlinear relationships.
- Works for classification and regression.
- Can model feature interactions.

### Limitations

- Can overfit easily.
- Small data changes may produce a different tree.
- Greedy splits do not guarantee the globally best tree.
- Deep trees may become difficult to interpret.
- Feature importance can be misleading when features are correlated.

### Exercise 2: Limit the tree

Train trees with `max_depth=2`, `max_depth=4`, and `max_depth=None`. Compare training accuracy, testing accuracy, and number of leaves.

````python
criteria = {
    "max_depth" : [n for n in range(2, 11, 2)],
    "max_leaf_nodes" : [n for n in range(2, 11, 2)],
    }
````

````python
# Your code here
accuracy_gini = []
accuracy_entropy = []

for n in criteria["max_depth"]:
    
    print("Max depth: {0}, Max Leaf Nodes: {1}".format(n, n))
    decision_tree_gini = DecisionTreeClassifier(criterion="gini", max_depth=n, max_leaf_nodes=n)
    decision_tree_gini.fit(iris_X, iris_y)
    decision_tree_entropy = DecisionTreeClassifier(criterion="entropy", max_depth=n, max_leaf_nodes=n)
    decision_tree_entropy.fit(iris_X, iris_y)
    accuracy_gini.append(decision_tree_gini.score(iris_X, iris_y))
    accuracy_entropy.append(decision_tree_entropy.score(iris_X, iris_y))

````

````python
print(f"Gini: {accuracy_gini} \nEntropy: {accuracy_entropy}")
````

### Exercise 3: New prediction

Create a new flower measurement and predict its class and class probabilities.

````python
# Your code here
print(iris_X[0].tolist())

array = np.array([[3, 3, 3, 1]])

pred_gini = decision_tree_gini.predict(array)
pred_entropy = decision_tree_entropy.predict(array)

prob_gini = decision_tree_gini.predict_proba(array)
prob_entropy = decision_tree_entropy.predict_proba(array)
````

````python
print(f"Gini: {pred_gini}, {prob_gini}")
print(f"Entropy: {pred_entropy}, {prob_entropy}")

print(iris.target_names[0])
````
